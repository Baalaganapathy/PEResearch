# PE License Growth Rate Analysis


This notebook explores the growth of Professional Engineer (PE) licenses using exam data.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
# Load combined PE exam data
df = pd.read_csv('../Data/combined_PE_data.csv')

cols = ['first_time_volume', 'first_time_pass_rate', 'repeat_volume', 'repeat_pass_rate', 'year']
for c in cols:
    df[c] = pd.to_numeric(df[c], errors='coerce')

# Total number of licenses granted (approximate passes)
df['passes'] = (df['first_time_volume'] * df['first_time_pass_rate'] / 100) + \
               (df['repeat_volume'] * df['repeat_pass_rate'] / 100)
```

## Overall Growth


```python
overall = df.groupby('year')['passes'].sum().reset_index()
overall['growth_rate'] = overall['passes'].pct_change() * 100

fig, ax1 = plt.subplots(figsize=(8, 4))
sns.lineplot(data=overall, x='year', y='passes', marker='o', ax=ax1, color='tab:blue')
ax1.set_title('Total PE Licenses Granted per Year')
ax1.set_ylabel('Number of licenses')
ax1.set_xlabel('Year')
plt.tight_layout()
plt.show()

fig, ax2 = plt.subplots(figsize=(8, 4))
sns.barplot(data=overall, x='year', y='growth_rate', color='tab:orange', ax=ax2)
ax2.set_title('Year-over-Year Growth Rate')
ax2.set_ylabel('Growth rate (%)')
ax2.set_xlabel('Year')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Growth by Discipline


```python
discipline = df.groupby(['year', 'pe_exam'])['passes'].sum().reset_index()
discipline['growth_rate'] = discipline.groupby('pe_exam')['passes'].pct_change() * 100
pivot = discipline.pivot(index='pe_exam', columns='year', values='growth_rate')

plt.figure(figsize=(12, 8))
sns.heatmap(pivot, cmap='RdBu', center=0, annot=True, fmt='.1f')
plt.title('Year-over-Year Growth Rate by Discipline (%)')
plt.xlabel('Year')
plt.ylabel('Discipline')
plt.tight_layout()
plt.show()
```
